# 1. Reasoning Task

## Fixed Test Input

A machine learning model makes 1,000 predictions.
It correctly predicts 920 cases.

What is the accuracy of the model?

## prompt_1_reasoning = 
A machine learning model makes 1,000 predictions.
It correctly predicts 920 cases.

What is the accuracy of the model?


result_1_reasoning =

$$
\text{Accuracy} = \frac{\text{Correct Predictions}}{\text{Total Predictions}} \times 100
$$

$$
= \frac{920}{1000} \times 100 = \boxed{92\%}
$$

✅ **Answer: 92% accuracy**


## prompt_2_reasoning = 
Calculate the accuracy of the machine learning model.

Given:
- Total predictions = 1,000
- Correct predictions = 920

Use this formula:
Accuracy = (Correct Predictions / Total Predictions) × 100

Show the calculation and give the final answer as a percentage.
Explain the result in one simple sentence.

The calculation is:

$$
\text{Accuracy} = \left(\frac{920}{1000}\right)\times100
$$

$$
= 0.92\times100
$$

$$
= \boxed{92\%}
$$

**Final Answer: 92%**

This means the model correctly predicted **92 out of every 100 cases**.


## Reasoning - Evaluation

### Baseline Prompt Weakness

The baseline prompt asks the question directly, but it does not specify
how the answer should be calculated or explained.

### Improvement

The improved prompt gives the formula, provides the values clearly, and
specifies that the calculation and final percentage should be shown.

### Evaluation Criteria

- Correctness
- Clear calculation
- Clear explanation
- Following instructions

### Outcome

The improved prompt gives more specific instructions and makes the expected
answer clearer.

# 4. Coding Task

## Fixed Test Input

Create Python code using pandas to calculate the average value of the
Score column in this dataset:

Name    Score
Alice   80
Bob     90
Charlie 70

## prompt_1_coding = Create Python code using pandas to calculate the average value of theScore column.

Data:
Name    Score
Alice   80
Bob     90
Charlie 70

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Score": [80, 90, 70]
}

df = pd.DataFrame(data)

average_score = df["Score"].mean()

print("Average Score:", average_score)
```
