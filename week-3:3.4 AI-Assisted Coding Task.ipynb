## 1. Short Specification

The goal is to create a small Python utility that reads a CSV file and provides a simple summary of the dataset.

The application should:

1. Load a CSV file using pandas.
2. Display the number of rows and columns.
3. Display the column names.
4. Display the number of missing values in each column.
5. Display basic numerical statistics.
6. Handle a missing or invalid file path without crashing.
7. Use simple and readable Python code.

The application should be small, understandable, and easy to test.


````python
import pandas as pd

data = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [20, 21, 19, 22],
    "Score": [85, 90, 78, 88]
})

data.to_csv("student_data.csv", index=False)
print("Test CSV file created successfully.")
````

## 3. AI-Generated Code

The initial implementation was generated with the help of an AI coding assistant based on the specification above.


````python
import os
import pandas as pd

file_path = "student_data.csv"

if os.path.exists(file_path):
    data = pd.read_csv(file_path)
    print("Number of rows:", data.shape[0])
    print("Number of columns:", data.shape[1])
    print("\nColumns found:")
    print(data.columns.tolist())
    print("\nMissing values:")
    print(data.isnull().sum())
    print("\nBasic statistics:")
    print(data.describe())
else:
    print("Error: File not found.")
````
